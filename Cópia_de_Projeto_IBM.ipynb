# **Introdução 2**

💼 Rotatividade em uma organização — Por que os funcionários pedem demissão?

Os funcionários são a espinha dorsal de uma organização. O desempenho da organização depende fortemente da qualidade de seus funcionários.

Os desafios enfrentados por uma organização devido à rotatividade de funcionários são:

Gastos de dinheiro e tempo para treinar novos funcionários.
Perda de funcionários experientes.
Impacto na produtividade.
Impacto no lucro.

Antes de começarmos a trabalhar com os dados, o primeiro passo é definir a pergunta de negócio. Ter clareza sobre as questões abaixo é fundamental, pois a solução desenvolvida só fará sentido se o problema estiver bem definido.

📌 Questões de negócio para reflexão
Quais fatores estão contribuindo mais para a rotatividade dos funcionários?
Quais medidas a empresa deve adotar para reter seus funcionários?
Qual valor o modelo traz para o negócio?
O modelo permitirá que a empresa economize dinheiro?
Quais áreas ou unidades de negócio enfrentam mais problemas de rotatividade?

```python
import requests
import pandas as pd

url = "https://apisidra.ibge.gov.br/values/t/5444/n3/all/v/5932/p/all/c694/all"

resposta = requests.get(url, timeout=30)
resposta.raise_for_status()

dados = resposta.json()

descricao_colunas = dados[0]

df = pd.DataFrame(dados[1:])

df = df.rename(columns=descricao_colunas)

df = df[
    [
        "Unidade da Federação",
        "Grupamentos ocupacionais no trabalho principal - PNADC",
        "Valor",
        "Unidade de Medida",
        "Trimestre"
    ]
]

df = df.rename(
    columns={
        "Unidade da Federação": "Estado",
        "Grupamentos ocupacionais no trabalho principal - PNADC": "Cargo",
        "Valor": "Salário médio",
        "Unidade de Medida": "Unidade",
        "Trimestre": "Período"
    }
)

df["Salário médio"] = pd.to_numeric(
    df["Salário médio"],
    errors="coerce"
)

df = df[df["Cargo"] != "Total"]

# Filtra somente 2022
df = df[df["Período"].str.contains("2022", na=False)]

# Mostra os 10 maiores salários
display(
    df.sort_values(
        "Salário médio",
        ascending=False
    ).head(10)
)
```

```python
# Criar faixas salariais

limites = [
    0,
    3000,
    5000,
    8000,
    12000,
    float("inf")
]

nomes_faixas = [
    "Até 3.000",
    "De 3.001 a 5.000",
    "De 5.001 a 8.000",
    "De 8.001 a 12.000",
    "Acima de 12.000"
]

df["FaixaSalarial"] = pd.cut(
    df["MonthlyIncome"],
    bins=limites,
    labels=nomes_faixas,
    include_lowest=True
)

display(
    df[
        ["EmployeeNumber", "MonthlyIncome", "FaixaSalarial"]
    ].head(10)
)
```

```python
distribuicao_salarial = (
    df["FaixaSalarial"]
    .value_counts()
    .sort_index()
    .rename_axis("Faixa salarial")
    .reset_index(name="Quantidade de funcionários")
)

display(distribuicao_salarial)
```

```python
distribuicao_salarial = (
    df["FaixaSalarial"]
    .value_counts()
    .sort_index()
    .rename_axis("Faixa salarial")
    .reset_index(name="Quantidade")
)

distribuicao_salarial["Percentual"] = (
    distribuicao_salarial["Quantidade"] / len(df) * 100
).round(2)

display(distribuicao_salarial)
```
